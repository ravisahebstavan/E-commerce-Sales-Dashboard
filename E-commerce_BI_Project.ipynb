```python
import pandas as pd # Import the pandas library, which is essential for data manipulation.

# Load the separate sheets from your single Excel file.
# We need to specify the sheet name for each year.
# Please check your Excel file to ensure these sheet names are correct.
df_2009_2010 = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2009-2010")
df_2010_2011 = pd.read_excel("online_retail_II.xlsx", sheet_name="Year 2010-2011")

# Combine the two DataFrames into a single, comprehensive dataset.
# 'ignore_index=True' ensures that the row numbering is continuous.
combined_df = pd.concat([df_2009_2010, df_2010_2011], ignore_index=True)

# Display the first few rows to confirm the data loaded and combined correctly.
print("Successfully loaded and combined the data.")
print("The combined dataset has", len(combined_df), "rows.")
print(combined_df.head())
```

### Data Cleaning and Preprocessing
The raw data often contains errors, missing values, or inconsistent formats. I will clean the data by:
-   Removing rows with missing Customer IDs, as they cannot be used for customer analysis.
-   Removing rows with negative quantities or prices, which represent returns and should be handled separately.
-   Creating a new column for the total price of each item.

```python
# Convert the 'InvoiceDate' column to a standard datetime format.
combined_df['InvoiceDate'] = pd.to_datetime(combined_df['InvoiceDate'])

# Remove rows with a missing 'Customer ID'
# The 'inplace=True' argument saves the changes directly to the dataframe.
combined_df.dropna(subset=['Customer ID'], inplace=True)

# Remove rows where 'Quantity' or 'Price' are less than or equal to zero.
combined_df = combined_df[(combined_df['Quantity'] > 0) & (combined_df['Price'] > 0)]

# Create a new column 'Total_Price' by multiplying 'Quantity' and 'Price'.
combined_df['Total_Price'] = combined_df['Quantity'] * combined_df['Price']

# RENAME THE COLUMN to remove the space, which causes SQL errors.
combined_df.rename(columns={'Customer ID': 'CustomerID'}, inplace=True)

# Print information to confirm the cleaning was successful.
print("After cleaning, the dataset has", len(combined_df), "rows.")
print("\nFirst 5 rows with the new 'Total_Price' column:")
print(combined_df.head())
```

```python
# Save the final, cleaned dataframe to a new CSV file.
# The 'index=False' argument prevents pandas from writing a row index column.
combined_df.to_csv("online_retail_cleaned.csv", index=False)

print("Successfully exported the cleaned data to 'online_retail_cleaned.csv'!")
```
